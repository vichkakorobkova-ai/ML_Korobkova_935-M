## Дано

````python
# Импорт библиотек
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, HuberRegressor, SGDRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

# ============================================================================
# Этап 1: Генерация и подготовка данных
# ============================================================================

print("\n" + "=" * 100)
print("                     Этап 1: Генерация и подготовка данных                     ")
print("=" * 100)

print("\n1. Генерация синтетического датасета\n")

# Использованы методы генерации синтетических данных из лекционного материала
# "Линейная регрессия и градиентный спуск"
def generate_dataset(n_samples=500, n_features=3, noise_std=0.8,  # σ=0.8 из варианта
                     outlier_fraction=0.08, outlier_multiplier=4,  # 8% выбросов, амплитуда ×4
                     random_seed=42):
    """
    Генерация синтетического датасета для линейной регрессии

    Параметры из варианта:
    - Целевая переменная: app_rating
    - Признаки: features, updates, support
    - Шум (σ) = 0.8
    - Процент выбросов = 8%
    - Амплитуда выбросов = ×4
    """
    if random_seed is not None:
        np.random.seed(random_seed)

    # Генерация признаков
    n_features = 3                        # features, updates, support
    X = np.random.randn(n_samples, n_features)

    # Истинные коэффициенты для признаков
    true_coef = np.array([2.5, 1.8, 3.2])  # Веса для features, updates, support
    true_intercept = 4.0                   # Базовый рейтинг

    # Генерация целевой переменной (app_rating)
    y = X @ true_coef + true_intercept + np.random.randn(n_samples) * noise_std

    print("\n2. Добавление шумов и выбросов\n")

    # Добавление выбросов (8% от данных)
    n_outliers = int(n_samples * outlier_fraction)
    outlier_idx = np.random.choice(n_samples, n_outliers, replace=False)
    y[outlier_idx] += np.random.choice([-1, 1], n_outliers) * \
                      np.random.uniform(3, 10, n_outliers) * outlier_multiplier

    print(f"Сгенерировано данных: {n_samples} наблюдений")
    print(f"Количество признаков: {n_features} (features, updates, support)")
    print(f"Добавлено выбросов: {n_outliers} ({outlier_fraction*100}%)")
    print(f"Истинные коэффициенты: {true_coef}")
    print(f"Истинный свободный член: {true_intercept}")

    return X, y, true_coef, true_intercept

# Генерация данных
X, y, true_coef, true_intercept = generate_dataset(
    n_samples=500,
    noise_std=0.8,        # σ = 0.8
    outlier_fraction=0.08, # 8% выбросов
    outlier_multiplier=4,  # амплитуда ×4
    random_seed=42
)

print("\n3. Разделение данных на обучающую и тестовую выборку\n")
# train_test_split из sklearn: https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html
# Разделение на обучающую и тестовую выборки (70/30)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print(f"Размер обучающей выборки: {X_train.shape[0]} наблюдений")
print(f"Размер тестовой выборки: {X_test.shape[0]} наблюдений")

print("\n4. Выполнение масштабирования признаков\n")
# StandardScaler: https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html
# Масштабирование признаков
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Масштабирование признаков выполнено (StandardScaler)")

# Создание DataFrame для удобства анализа
feature_names = ['features', 'updates', 'support']
df = pd.DataFrame(X, columns=feature_names)
df['app_rating'] = y

print("\nПервые 10 строк датасета:")
print(df.head(10))


# ============================================================================
# Этап 2: Разведочный анализ данных (EDA)
# ============================================================================

print("\n" + "=" * 100)
print("                     Этап 2: Разведочный анализ данных (EDA)                     ")
print("=" * 100)

# Методы визуализации: seaborn.pairplot, matplotlib
# Документация seaborn: https://seaborn.pydata.org/

# Создаем фигуру для нескольких графиков
plt.figure(figsize=(15, 10))

print("1. Построение матрицы корреляции признаков (heatmap)\n")

# 1. Матрица корреляций (heatmap)
plt.subplot(2, 2, 1)
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": 0.8},
            annot_kws={'size': 10, 'weight': 'bold'})
plt.title('Матрица корреляций признаков', fontsize=14, fontweight='bold')

# Текстовый вывод корреляций
print("Корреляционная матрица:")
print(correlation_matrix.round(3))
print("\nНаиболее сильные корреляции с целевой переменной:")
for feat in feature_names:
    print(f"  {feat} vs app_rating: {correlation_matrix.loc['app_rating', feat]:.3f}")

print("\n2. Создание pairplot для визуализации линейных зависимостей\n")

# Создаем pairplot отдельно
pairplot_fig = sns.pairplot(df, diag_kind='hist', plot_kws={'alpha': 0.6})
pairplot_fig.fig.suptitle('Pairplot признаков и целевой переменной',
                          y=1.02, fontsize=14, fontweight='bold')
plt.show()

print("3. Анализ распределения целевой переменной\n")

plt.figure(figsize=(15, 10))  # Создание новой фигуры для оставшихся графиков
plt.subplot(2, 2, 1)
sns.histplot(df['app_rating'], kde=True, bins=30, color='skyblue')
plt.axvline(df['app_rating'].mean(), color='red', linestyle='--', linewidth=2,
            label=f"Среднее: {df['app_rating'].mean():.2f}")
plt.axvline(df['app_rating'].median(), color='green', linestyle='--', linewidth=2,
            label=f"Медиана: {df['app_rating'].median():.2f}")
plt.axvline(df['app_rating'].mode().iloc[0] if not df['app_rating'].mode().empty else df['app_rating'].mean(),
            color='purple', linestyle='--', linewidth=2,
            label=f"Мода: {df['app_rating'].mode().iloc[0]:.2f}" if not df['app_rating'].mode().empty else "Мода не определена")
plt.title('Распределение целевой переменной (app_rating)', fontsize=14, fontweight='bold')
plt.xlabel('app_rating')
plt.ylabel('Частота')
plt.legend()
plt.grid(alpha=0.3)

# Добавим статистику распределения
print("Статистика распределения app_rating:")
print(f"  Среднее: {df['app_rating'].mean():.3f}")
print(f"  Медиана: {df['app_rating'].median():.3f}")
print(f"  Мода: {df['app_rating'].mode().iloc[0]:.3f}" if not df['app_rating'].mode().empty else "  Мода: не определена")
print(f"  Стандартное отклонение: {df['app_rating'].std():.3f}")
print(f"  Минимум: {df['app_rating'].min():.3f}")
print(f"  Максимум: {df['app_rating'].max():.3f}")
print(f"  Асимметрия (skewness): {df['app_rating'].skew():.3f}")
print(f"  Эксцесс (kurtosis): {df['app_rating'].kurtosis():.3f}")

print("\n4. Выявление и визуализация выбросов\n")

plt.subplot(2, 2, 2)
df_box = df.copy()
df_box_melted = pd.melt(df_box, value_vars=feature_names + ['app_rating'],
                        var_name='Переменная', value_name='Значение')
# Исправлено: используем hue вместо palette
sns.boxplot(x='Переменная', y='Значение', hue='Переменная', data=df_box_melted,
            palette='Set2', legend=False)
plt.title('Boxplot признаков и целевой переменной', fontsize=14, fontweight='bold')
plt.xlabel('')
plt.ylabel('Значения')
plt.xticks(rotation=45)
plt.grid(alpha=0.3, axis='y')

# Добавим scatter plot для визуализации выбросов
plt.subplot(2, 2, 3)
for i, feat in enumerate(feature_names):
    plt.scatter(df[feat], df['app_rating'], alpha=0.5, label=feat, s=30)
plt.xlabel('Значения признаков')
plt.ylabel('app_rating')
plt.title('Зависимость app_rating от признаков', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(alpha=0.3)

# Добавим violin plot для более детального анализа распределений
plt.subplot(2, 2, 4)
df_violin = df.copy()
df_violin_melted = pd.melt(df_violin, value_vars=feature_names + ['app_rating'],
                           var_name='Переменная', value_name='Значение')
# Исправлено: используем hue вместо palette
sns.violinplot(x='Переменная', y='Значение', hue='Переменная', data=df_violin_melted,
               palette='Set3', legend=False)
plt.title('Violin plot признаков и целевой переменной', fontsize=14, fontweight='bold')
plt.xlabel('')
plt.ylabel('Значения')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

# Дополнительный анализ выбросов
Q1 = df['app_rating'].quantile(0.25)
Q3 = df['app_rating'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['app_rating'] < lower_bound) | (df['app_rating'] > upper_bound)]

print("\n" + "="*60)
print("ДЕТАЛЬНЫЙ АНАЛИЗ ВЫБРОСОВ:")
print("="*60)
print(f"Нижняя граница (Q1 - 1.5*IQR): {lower_bound:.2f}")
print(f"Верхняя граница (Q3 + 1.5*IQR): {upper_bound:.2f}")
print(f"Количество выбросов по методу межквартильного размаха: {len(outliers)}")
print(f"Процент выбросов: {len(outliers)/len(df)*100:.2f}%")

if len(outliers) > 0:
    print("\nСтатистика по выбросам:")
    print(f"  Минимальное значение среди выбросов: {outliers['app_rating'].min():.2f}")
    print(f"  Максимальное значение среди выбросов: {outliers['app_rating'].max():.2f}")
    print(f"  Среднее значение среди выбросов: {outliers['app_rating'].mean():.2f}")

# Визуализация выбросов на фоне распределения
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(range(len(df)), df['app_rating'], alpha=0.6, c='blue', label='Нормальные значения')
plt.scatter(outliers.index, outliers['app_rating'], alpha=0.8, c='red', s=80,
            label='Выбросы', edgecolors='black', linewidth=1)
plt.axhline(y=lower_bound, color='green', linestyle='--', label=f'Нижняя граница: {lower_bound:.2f}')
plt.axhline(y=upper_bound, color='orange', linestyle='--', label=f'Верхняя граница: {upper_bound:.2f}')
plt.xlabel('Индекс наблюдения')
plt.ylabel('app_rating')
plt.title('Визуализация выбросов в целевой переменной', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
# Исправлено: используем boxplot без palette
sns.boxplot(x=df['app_rating'], color='lightblue')
# Исправлено: используем scatter вместо swarmplot для наглядности
outliers_x = outliers['app_rating'].values
plt.scatter(outliers_x, [0.1] * len(outliers_x), color='red', s=100,
            label='Выбросы', alpha=0.7, edgecolors='black', linewidth=1, zorder=5)
plt.xlabel('app_rating')
plt.title('Boxplot с выделенными выбросами', fontsize=14, fontweight='bold')
plt.legend()

plt.tight_layout()
plt.show()

# Добавим еще один полезный график - распределение остатков после удаления выбросов
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.histplot(df['app_rating'], kde=True, color='blue', alpha=0.5, label='Все данные')
sns.histplot(df[~df.index.isin(outliers.index)]['app_rating'],
             kde=True, color='green', alpha=0.5, label='Без выбросов')
plt.axvline(df['app_rating'].mean(), color='blue', linestyle='--', label=f'Среднее (все): {df["app_rating"].mean():.2f}')
plt.axvline(df[~df.index.isin(outliers.index)]['app_rating'].mean(),
            color='green', linestyle='--', label=f'Среднее (без выбросов): {df[~df.index.isin(outliers.index)]["app_rating"].mean():.2f}')
plt.title('Сравнение распределений с выбросами и без', fontsize=14, fontweight='bold')
plt.xlabel('app_rating')
plt.ylabel('Частота')
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
# Q-Q plot для проверки нормальности
from scipy import stats
stats.probplot(df['app_rating'], dist="norm", plot=plt)
plt.title('Q-Q plot (проверка нормальности)', fontsize=14, fontweight='bold')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n" + "="*60)
print("ВЫВОДЫ ПО ЭТАПУ 2: Разведочный анализ данных (EDA):")
print("="*60)
print("1. Корреляционный анализ:")
for feat in feature_names:
    corr = correlation_matrix.loc['app_rating', feat]
    strength = "сильная" if abs(corr) > 0.7 else "умеренная" if abs(corr) > 0.3 else "слабая"
    direction = "положительная" if corr > 0 else "отрицательная"
    print(f"   - {feat}: {direction} {strength} корреляция ({corr:.3f})")

print(f"\n2. Характеристики распределения app_rating:")
skewness = df['app_rating'].skew()
if abs(skewness) < 0.5:
    skew_desc = "симметричное"
elif abs(skewness) < 1:
    skew_desc = "умеренно асимметричное"
else:
    skew_desc = "сильно асимметричное"
print(f"   - Распределение: {skew_desc} (коэф. асимметрии = {skewness:.3f})")
print(f"   - Наличие выбросов: {'да' if len(outliers) > 0 else 'нет'} ({len(outliers)} наблюдений, {len(outliers)/len(df)*100:.1f}%)")
print(f"   - Размах значений: от {df['app_rating'].min():.2f} до {df['app_rating'].max():.2f}")

# ============================================================================
# Этап 3: Обучение моделей
# ============================================================================

print("\n" + "=" * 100)
print("                         Этап 3: Обучение моделей                         ")
print("=" * 100)

# LinearRegression: https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html
# HuberRegressor: https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.HuberRegressor.html
# SGDRegressor: https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.SGDRegressor.html

print("""1. Обучите три модели
""")
# 1. LinearRegression (классическая линейная регрессия)
model_lr = LinearRegression()
model_lr.fit(X_train_scaled, y_train)
print("\n1.1. LinearRegression (минимизация MSE):")
print(f"   Коэффициенты: {model_lr.coef_}")
print(f"   Свободный член: {model_lr.intercept_:.4f}")

# 2. HuberRegressor (робастная регрессия)
model_huber = HuberRegressor(epsilon=2.0, max_iter=1000)  # epsilon = 2.0 из варианта
model_huber.fit(X_train_scaled, y_train)
print("\n1.2. HuberRegressor (epsilon=2.0):")
print(f"   Коэффициенты: {model_huber.coef_}")
print(f"   Свободный член: {model_huber.intercept_:.4f}")

# 3. SGDRegressor с функцией потерь huber
model_sgd = SGDRegressor(
    loss='huber',           # loss = huber из варианта
    epsilon=2.0,            # тот же epsilon для совместимости
    learning_rate='constant',
    eta0=0.005,             # learning rate = 0.005 из варианта
    max_iter=1600,          # итерации = 1600 из варианта
    random_state=42,
    tol=1e-3
)

model_sgd.fit(X_train_scaled, y_train)
print("\n1.3. SGDRegressor (loss=huber, lr=0.005, iter=1600):")
print(f"   Коэффициенты: {model_sgd.coef_}")
print(f"   Свободный член: {model_sgd.intercept_[0]:.4f}")

# Сравнение с истинными значениями
print("\n" + " "*30)
print("СРАВНЕНИЕ С ИСТИННЫМИ КОЭФФИЦИЕНТАМИ:")
print(f"Истинные коэффициенты: {true_coef}")
print(f"Истинный свободный член: {true_intercept:.4f}")


# ============================================================================
# Этап 4: Оценка качества
# ============================================================================

print("\n" + "=" * 100)
print("                         Этап 4: Оценка качества                         ")
print("=" * 100)

# Метрики качества:
# mean_squared_error: https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_squared_error.html
# mean_absolute_error: https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html
# r2_score: https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.htm

# Определяем функцию сравнения моделей
def compare_models(models_dict, X_test, y_test):
    """
    Сравнение нескольких моделей регрессии по метрикам качества

    Parameters:
    -----------
    models_dict : dict
        Словарь вида {'имя_модели': модель, ...}
    X_test, y_test : array-like
        Тестовые данные

    Returns:
    --------
    pd.DataFrame
        DataFrame с метриками MSE, MAE, R2 для всех моделей
    """
    results = []
    for name, model in models_dict.items():
        y_pred = model.predict(X_test)
        results.append({
            'Model': name,
            'MSE': mean_squared_error(y_test, y_pred),
            'MAE': mean_absolute_error(y_test, y_pred),
            'R2': r2_score(y_test, y_pred)
        })
    return pd.DataFrame(results)

print("\n1. Расчет метрик на тестовой выборке: MSE, MAE, R²\n")

# Создаем словарь с обученными моделями
models_dict = {
    'Linear Regression': model_lr,
    'Huber Regressor (ε=2.0)': model_huber,
    'SGD Regressor (huber)': model_sgd
}

# Используем функцию для сравнения моделей
results_df = compare_models(models_dict, X_test_scaled, y_test)

# Добавляем колонку с RMSE (квадратный корень из MSE)
results_df['RMSE'] = np.sqrt(results_df['MSE'])

# Переупорядочиваем колонки для лучшей читаемости
results_df = results_df[['Model', 'MSE', 'RMSE', 'MAE', 'R2']]

print("\n" + "=" * 60)
print(" ТАБЛИЦА МЕТРИК КАЧЕСТВА МОДЕЛЕЙ")
print("=" * 60)
print(results_df.round(4).to_string(index=False))

# Определение лучшей модели по каждой метрике
print("\n" + "-" * 60)
print(" Анализ метрик:")
print("-" * 60)

best_mse = results_df.loc[results_df['MSE'].idxmin(), 'Model']
best_rmse = results_df.loc[results_df['RMSE'].idxmin(), 'Model']
best_mae = results_df.loc[results_df['MAE'].idxmin(), 'Model']
best_r2 = results_df.loc[results_df['R2'].idxmax(), 'Model']

print(f"    Лучшая модель по MSE: {best_mse} (MSE = {results_df['MSE'].min():.4f})")
print(f"    Лучшая модель по RMSE: {best_rmse} (RMSE = {results_df['RMSE'].min():.4f})")
print(f"    Лучшая модель по MAE: {best_mae} (MAE = {results_df['MAE'].min():.4f})")
print(f"    Лучшая модель по R²: {best_r2} (R² = {results_df['R2'].max():.4f})")

print("\n2. Построение графиков (фактические vs предсказанные значения)\n")

# Получаем предсказания для всех моделей
y_preds = {}
for name, model in models_dict.items():
    y_preds[name] = model.predict(X_test_scaled)

# Создание графиков фактические vs предсказанные значения
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, (name, y_pred) in enumerate(y_preds.items()):
    ax = axes[idx]

    # Scatter plot
    ax.scatter(y_test, y_pred, alpha=0.6, edgecolors='black', linewidth=0.5, s=50)

    # Линия идеального предсказания
    min_val = min(y_test.min(), y_pred.min())
    max_val = max(y_test.max(), y_pred.max())
    ax.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2, label='Идеальное предсказание')

    # Линия регрессии (аппроксимация)
    z = np.polyfit(y_test, y_pred, 1)
    p = np.poly1d(z)
    ax.plot([min_val, max_val], p([min_val, max_val]), 'b-', linewidth=1.5, alpha=0.7, label='Линия регрессии')

    # Получаем метрики для этой модели из results_df
    model_metrics = results_df[results_df['Model'] == name].iloc[0]

    # Добавляем текст с метриками
    ax.text(0.05, 0.95, f'R² = {model_metrics["R2"]:.4f}\nMAE = {model_metrics["MAE"]:.4f}\nRMSE = {model_metrics["RMSE"]:.4f}',
            transform=ax.transAxes, fontsize=10, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

    ax.set_xlabel('Фактические значения', fontsize=11)
    ax.set_ylabel('Предсказанные значения', fontsize=11)
    ax.set_title(f'{name}', fontsize=13, fontweight='bold')
    ax.legend(loc='lower right')
    ax.grid(alpha=0.3)
    ax.axis('equal')

plt.suptitle('Фактические vs Предсказанные значения для трех моделей', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("\n3. Остатки модели (residual plot)\n")

# Создание residual plots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, (name, y_pred) in enumerate(y_preds.items()):
    ax = axes[idx]
    residuals = y_test - y_pred

    # Scatter plot остатков
    ax.scatter(y_pred, residuals, alpha=0.6, edgecolors='black', linewidth=0.5, s=50)
    ax.axhline(y=0, color='r', linestyle='--', linewidth=2, label='Нулевая линия')

    # Добавляем скользящее среднее для выявления паттернов
    from scipy.ndimage import uniform_filter1d
    sorted_idx = np.argsort(y_pred)
    sorted_pred = y_pred[sorted_idx]
    sorted_resid = residuals[sorted_idx]

    # Сглаживание остатков
    smoothed = uniform_filter1d(sorted_resid, size=20)
    ax.plot(sorted_pred, smoothed, 'g-', linewidth=2, label='Скользящее среднее')

    # Добавляем текст со статистикой остатков
    std_resid = np.std(residuals)
    mean_resid = np.mean(residuals)
    ax.text(0.05, 0.95, f'Ср. остатков: {mean_resid:.4f}\nСтд. остатков: {std_resid:.4f}',
            transform=ax.transAxes, fontsize=10, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))

    ax.set_xlabel('Предсказанные значения', fontsize=11)
    ax.set_ylabel('Остатки', fontsize=11)
    ax.set_title(f'Residual Plot - {name}', fontsize=13, fontweight='bold')
    ax.legend(loc='upper right')
    ax.grid(alpha=0.3)

plt.suptitle('Анализ остатков моделей', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# Дополнительный анализ остатков (гистограммы)
fig, axes = plt.subplots(1, 3, figsize=(18, 4))

for idx, (name, y_pred) in enumerate(y_preds.items()):
    ax = axes[idx]
    residuals = y_test - y_pred

    # Гистограмма остатков
    ax.hist(residuals, bins=30, alpha=0.7, color='steelblue', edgecolor='black', density=True)

    # Добавляем нормальное распределение для сравнения
    from scipy.stats import norm
    x = np.linspace(residuals.min(), residuals.max(), 100)
    ax.plot(x, norm.pdf(x, np.mean(residuals), np.std(residuals)),
            'r-', linewidth=2, label='Нормальное распределение')

    ax.axvline(x=0, color='green', linestyle='--', linewidth=2, label='Нулевая линия')
    ax.set_xlabel('Остатки', fontsize=11)
    ax.set_ylabel('Плотность', fontsize=11)
    ax.set_title(f'Распределение остатков - {name}', fontsize=13, fontweight='bold')
    ax.legend()
    ax.grid(alpha=0.3)

plt.suptitle('Гистограммы распределения остатков', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("\n4. Сравнение предсказаний трёх моделей\n")

# График сравнения предсказаний всех моделей
plt.figure(figsize=(15, 10))

# График 1: Сравнение предсказаний на тестовой выборке
plt.subplot(2, 2, 1)
x_plot = np.arange(len(y_test[:50]))  # Первые 50 наблюдений для наглядности

plt.plot(x_plot, y_test[:50], 'ko-', linewidth=2, markersize=6, label='Фактические', alpha=0.8)
plt.plot(x_plot, y_preds['Linear Regression'][:50], 's-', linewidth=1.5, markersize=5, label='Linear', alpha=0.7)
plt.plot(x_plot, y_preds['Huber Regressor (ε=2.0)'][:50], '^-', linewidth=1.5, markersize=5, label='Huber', alpha=0.7)
plt.plot(x_plot, y_preds['SGD Regressor (huber)'][:50], 'D-', linewidth=1.5, markersize=5, label='SGD', alpha=0.7)

plt.xlabel('Индекс наблюдения', fontsize=12)
plt.ylabel('app_rating', fontsize=12)
plt.title('Сравнение предсказаний моделей (первые 50 наблюдений)', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(alpha=0.3)

# График 2: Bar plot для сравнения метрик
plt.subplot(2, 2, 2)
metrics_for_plot = results_df.melt(id_vars=['Model'],
                                    value_vars=['MSE', 'MAE', 'R2'],
                                    var_name='Metric', value_name='Value')

# Нормализация R2 для совместного отображения (R2 обычно от 0 до 1)
metrics_for_plot.loc[metrics_for_plot['Metric'] == 'R2', 'Value'] *= 100

sns.barplot(data=metrics_for_plot, x='Model', y='Value', hue='Metric', palette='Set2')
plt.title('Сравнение метрик моделей\n(R2 умножен на 100 для масштабирования)',
          fontsize=14, fontweight='bold')
plt.xlabel('Модель', fontsize=12)
plt.ylabel('Значение метрики', fontsize=12)
plt.legend(title='Метрика')
plt.xticks(rotation=15)
plt.grid(alpha=0.3, axis='y')

# График 3: Распределение абсолютных ошибок
plt.subplot(2, 2, 3)
errors_lr = np.abs(y_test - y_preds['Linear Regression'])
errors_huber = np.abs(y_test - y_preds['Huber Regressor (ε=2.0)'])
errors_sgd = np.abs(y_test - y_preds['SGD Regressor (huber)'])

# Проверка версии matplotlib для совместимости
import matplotlib
from packaging import version

if version.parse(matplotlib.__version__) >= version.parse('3.9.0'):
    bp = plt.boxplot([errors_lr, errors_huber, errors_sgd],
                     tick_labels=['Linear', 'Huber', 'SGD'],
                     patch_artist=True,
                     boxprops=dict(alpha=0.7),
                     medianprops=dict(color='red', linewidth=2))
else:
    bp = plt.boxplot([errors_lr, errors_huber, errors_sgd],
                     labels=['Linear', 'Huber', 'SGD'],
                     patch_artist=True,
                     boxprops=dict(alpha=0.7),
                     medianprops=dict(color='red', linewidth=2))

# Настройка цветов для ящиков
colors = ['lightblue', 'lightgreen', 'lightcoral']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)

plt.ylabel('Абсолютная ошибка |y_true - y_pred|', fontsize=12)
plt.title('Распределение абсолютных ошибок предсказаний', fontsize=14, fontweight='bold')
plt.grid(alpha=0.3, axis='y')

# Добавляем средние значения
means = [np.mean(errors_lr), np.mean(errors_huber), np.mean(errors_sgd)]
for i, mean_val in enumerate(means):
    plt.text(i+1, mean_val + 0.1, f'Ср.: {mean_val:.3f}',
             ha='center', va='bottom', fontsize=9, fontweight='bold')

# График 4: Сравнение коэффициентов детерминации
plt.subplot(2, 2, 4)
models_short = ['Linear', 'Huber', 'SGD']
r2_values = results_df['R2'].values
colors_r2 = ['skyblue', 'lightgreen', 'salmon']

bars = plt.bar(models_short, r2_values, color=colors_r2, alpha=0.7, edgecolor='black')
plt.axhline(y=results_df['R2'].mean(), color='red', linestyle='--', label=f'Средний R²: {results_df["R2"].mean():.4f}')
plt.ylabel('R² Score', fontsize=12)
plt.title('Сравнение коэффициентов детерминации R²', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(alpha=0.3, axis='y')

# Добавляем значения на столбцы
for bar, val in zip(bars, r2_values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
             f'{val:.4f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n" + "=" * 60)
print("ВЫВОДЫ ПО ЭТАПУ 4: ОЦЕНКА КАЧЕСТВА ")
print("=" * 60)

# Анализ результатов
print(f"\n1. Сравнение качества моделей:")
for _, row in results_df.iterrows():
    emoji = "1" if row['R2'] == results_df['R2'].max() else "2" if row['R2'] == sorted(results_df['R2'])[-2] else "3"
    print(f"   {emoji} {row['Model']}: R²={row['R2']:.4f}, MSE={row['MSE']:.4f}, MAE={row['MAE']:.4f}")

print(f"\n2. Анализ остатков:")
for name, y_pred in y_preds.items():
    residuals = y_test - y_pred
    mean_resid = np.mean(residuals)
    std_resid = np.std(residuals)

    # Проверка на систематическую ошибку
    if abs(mean_resid) < 0.1:
        bias_text = "отсутствует систематическая ошибка"
    else:
        bias_text = f"есть смещение ({mean_resid:.4f})"

    print(f"    {name}: среднее={mean_resid:.4f}, стд={std_resid:.4f} - {bias_text}")

# ============================================================================
# ЭТАП 5: Анализ устойчивости (исправленная версия)
# ============================================================================

print("\n" + "="*50)
print("ЭТАП 5: АНАЛИЗ УСТОЙЧИВОСТИ МОДЕЛЕЙ")
print("="*50)

# Адаптировано из лекционного материала "Линейная регрессия и градиентный спуск"
# Методы анализа выбросов основаны на межквартильном размахе (IQR)

# 1. Исследование влияния выбросов
# Создадим версию данных без выбросов для сравнения
outlier_indices = df[(df['app_rating'] < lower_bound) | (df['app_rating'] > upper_bound)].index
X_clean = np.delete(X, outlier_indices, axis=0)
y_clean = np.delete(y, outlier_indices, axis=0)

print(f"\nКоличество наблюдений в исходных данных: {len(X)}")
print(f"Количество выбросов: {len(outlier_indices)}")
print(f"Количество наблюдений в чистых данных: {len(X_clean)}")

# Разделение и масштабирование чистых данных
X_train_clean, X_test_clean, y_train_clean, y_test_clean = train_test_split(
    X_clean, y_clean, test_size=0.3, random_state=42
)

# Создаем новый scaler для чистых данных, чтобы избежать влияния выбросов на масштабирование
scaler_clean = StandardScaler()
X_train_clean_scaled = scaler_clean.fit_transform(X_train_clean)
X_test_clean_scaled = scaler_clean.transform(X_test_clean)

# Обучение на чистых данных
model_lr_clean = LinearRegression()
model_lr_clean.fit(X_train_clean_scaled, y_train_clean)

model_huber_clean = HuberRegressor(epsilon=2.0, max_iter=1000)
model_huber_clean.fit(X_train_clean_scaled, y_train_clean)

# Сравнение коэффициентов
print("\n" + "="*60)
print("ВЛИЯНИЕ ВЫБРОСОВ НА КОЭФФИЦИЕНТЫ МОДЕЛЕЙ")
print("="*60)
print(f"{'Модель':<20} {'Условие':<15} {'features':<12} {'updates':<12} {'support':<12}")
print("-" * 60)

# Linear Regression
print(f"{'Linear':<20} {'С выбросами':<15} {model_lr.coef_[0]:<12.4f} {model_lr.coef_[1]:<12.4f} {model_lr.coef_[2]:<12.4f}")
print(f"{'Linear':<20} {'Без выбросов':<15} {model_lr_clean.coef_[0]:<12.4f} {model_lr_clean.coef_[1]:<12.4f} {model_lr_clean.coef_[2]:<12.4f}")
print(f"{'Linear':<20} {'Разница':<15} {abs(model_lr.coef_[0]-model_lr_clean.coef_[0]):<12.4f} {abs(model_lr.coef_[1]-model_lr_clean.coef_[1]):<12.4f} {abs(model_lr.coef_[2]-model_lr_clean.coef_[2]):<12.4f}")

# Huber Regressor
print(f"\n{'Huber':<20} {'С выбросами':<15} {model_huber.coef_[0]:<12.4f} {model_huber.coef_[1]:<12.4f} {model_huber.coef_[2]:<12.4f}")
print(f"{'Huber':<20} {'Без выбросов':<15} {model_huber_clean.coef_[0]:<12.4f} {model_huber_clean.coef_[1]:<12.4f} {model_huber_clean.coef_[2]:<12.4f}")
print(f"{'Huber':<20} {'Разница':<15} {abs(model_huber.coef_[0]-model_huber_clean.coef_[0]):<12.4f} {abs(model_huber.coef_[1]-model_huber_clean.coef_[1]):<12.4f} {abs(model_huber.coef_[2]-model_huber_clean.coef_[2]):<12.4f}")

# 2. Влияние параметра epsilon на HuberRegressor
epsilon_values = np.linspace(1.0, 3.5, 15)  # Увеличим количество точек для более плавного графика
huber_metrics = {'epsilon': [], 'mse': [], 'mae': [], 'r2': []}

for eps in epsilon_values:
    model = HuberRegressor(epsilon=eps, max_iter=1000)
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)

    huber_metrics['epsilon'].append(eps)
    huber_metrics['mse'].append(mean_squared_error(y_test, y_pred))
    huber_metrics['mae'].append(mean_absolute_error(y_test, y_pred))
    huber_metrics['r2'].append(r2_score(y_test, y_pred))

# Визуализация влияния epsilon
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# График MSE
axes[0].plot(huber_metrics['epsilon'], huber_metrics['mse'], 'o-', linewidth=2, markersize=6, color='blue')
axes[0].axvline(x=2.0, color='red', linestyle='--', linewidth=2, label='ε=2.0 (заданный)')
opt_eps_mse = epsilon_values[np.argmin(huber_metrics['mse'])]
axes[0].axvline(x=opt_eps_mse, color='green', linestyle=':', linewidth=2, label=f'Оптимум ε={opt_eps_mse:.2f}')
axes[0].set_xlabel('ε (epsilon)', fontsize=12)
axes[0].set_ylabel('MSE', fontsize=12)
axes[0].set_title('Влияние ε на MSE', fontsize=14, fontweight='bold')
axes[0].legend()
axes[0].grid(alpha=0.3)

# График MAE
axes[1].plot(huber_metrics['epsilon'], huber_metrics['mae'], 'o-', linewidth=2, markersize=6, color='green')
axes[1].axvline(x=2.0, color='red', linestyle='--', linewidth=2, label='ε=2.0 (заданный)')
opt_eps_mae = epsilon_values[np.argmin(huber_metrics['mae'])]
axes[1].axvline(x=opt_eps_mae, color='blue', linestyle=':', linewidth=2, label=f'Оптимум ε={opt_eps_mae:.2f}')
axes[1].set_xlabel('ε (epsilon)', fontsize=12)
axes[1].set_ylabel('MAE', fontsize=12)
axes[1].set_title('Влияние ε на MAE', fontsize=14, fontweight='bold')
axes[1].legend()
axes[1].grid(alpha=0.3)

# График R²
axes[2].plot(huber_metrics['epsilon'], huber_metrics['r2'], 'o-', linewidth=2, markersize=6, color='orange')
axes[2].axvline(x=2.0, color='red', linestyle='--', linewidth=2, label='ε=2.0 (заданный)')
opt_eps_r2 = epsilon_values[np.argmax(huber_metrics['r2'])]
axes[2].axvline(x=opt_eps_r2, color='purple', linestyle=':', linewidth=2, label=f'Оптимум ε={opt_eps_r2:.2f}')
axes[2].set_xlabel('ε (epsilon)', fontsize=12)
axes[2].set_ylabel('R²', fontsize=12)
axes[2].set_title('Влияние ε на R²', fontsize=14, fontweight='bold')
axes[2].legend()
axes[2].grid(alpha=0.3)

plt.suptitle('Анализ чувствительности HuberRegressor к параметру epsilon',
             fontsize=16, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()

print("\n" + "=" * 60)
print("ВЫВОДЫ ПО ЭТАПУ 5: АНАЛИЗ УСТОЙЧИВОСТИ МОДЕЛЕЙ")
print("=" * 60)

# Анализ результатов
print("\n 1. Влияние выбросов на качество моделей:")
print("-" * 60)

# Получаем предсказания на чистых данных
y_pred_lr_clean = model_lr_clean.predict(X_test_clean_scaled)
y_pred_huber_clean = model_huber_clean.predict(X_test_clean_scaled)

# Расчет метрик для чистых данных
mse_lr_clean = mean_squared_error(y_test_clean, y_pred_lr_clean)
r2_lr_clean = r2_score(y_test_clean, y_pred_lr_clean)
mse_huber_clean = mean_squared_error(y_test_clean, y_pred_huber_clean)
r2_huber_clean = r2_score(y_test_clean, y_pred_huber_clean)

# Значения на данных с выбросами
lr_r2_with_outliers = results_df[results_df['Model'] == 'Linear Regression']['R2'].values[0]
huber_r2_with_outliers = results_df[results_df['Model'] == 'Huber Regressor (ε=2.0)']['R2'].values[0]

print(f"\n   Линейная регрессия:")
print(f"     - С выбросами: R² = {lr_r2_with_outliers:.4f}")
print(f"     - Без выбросов: R² = {r2_lr_clean:.4f}")
print(f"     - Изменение R²: +{r2_lr_clean - lr_r2_with_outliers:.4f}")
print(f"     - Улучшение качества: {(r2_lr_clean - lr_r2_with_outliers)/abs(lr_r2_with_outliers)*100:.1f}%")

print(f"\n   Huber Regressor:")
print(f"     - С выбросами: R² = {huber_r2_with_outliers:.4f}")
print(f"     - Без выбросов: R² = {r2_huber_clean:.4f}")
print(f"     - Изменение R²: +{r2_huber_clean - huber_r2_with_outliers:.4f}")
print(f"     - Улучшение качества: {(r2_huber_clean - huber_r2_with_outliers)/abs(huber_r2_with_outliers)*100:.1f}%")

# Расчет устойчивости коэффициентов
lr_coef_diff = np.mean(np.abs(model_lr.coef_ - model_lr_clean.coef_))
huber_coef_diff = np.mean(np.abs(model_huber.coef_ - model_huber_clean.coef_))

print(f"\n 2. Устойчивость коэффициентов к выбросам:")
print("-" * 60)
print(f"   Среднее изменение коэффициентов LinearRegression: {lr_coef_diff:.4f}")
print(f"   Среднее изменение коэффициентов HuberRegressor: {huber_coef_diff:.4f}")
print(f"   HuberRegressor устойчивее в {lr_coef_diff/huber_coef_diff:.1f} раз")

print(f"\n 3. Анализ параметра epsilon для HuberRegressor:")
print("-" * 60)

# Находим оптимальные значения epsilon
opt_eps_mse = epsilon_values[np.argmin(huber_metrics['mse'])]
opt_eps_mae = epsilon_values[np.argmin(huber_metrics['mae'])]
opt_eps_r2 = epsilon_values[np.argmax(huber_metrics['r2'])]

print(f"\n   • Оптимальное epsilon по MSE: {opt_eps_mse:.2f} (MSE = {min(huber_metrics['mse']):.4f})")
print(f"   • Оптимальное epsilon по MAE: {opt_eps_mae:.2f} (MAE = {min(huber_metrics['mae']):.4f})")
print(f"   • Оптимальное epsilon по R²: {opt_eps_r2:.2f} (R² = {max(huber_metrics['r2']):.4f})")

# Исправление: находим ближайшее значение к 2.0, а не точное совпадение
eps_array = np.array(huber_metrics['epsilon'])
idx_closest_to_2 = np.argmin(np.abs(eps_array - 2.0))
closest_eps = eps_array[idx_closest_to_2]

print(f"\n  Ближайшее значение к заданному epsilon=2.0: {closest_eps:.2f}")
print(f"     - MSE: {huber_metrics['mse'][idx_closest_to_2]:.4f}")
print(f"     - MAE: {huber_metrics['mae'][idx_closest_to_2]:.4f}")
print(f"     - R²: {huber_metrics['r2'][idx_closest_to_2]:.4f}")

# Сравнение с оптимальным значением
print(f"\n  Сравнение с оптимальным значением ({opt_eps_r2:.2f}):")
print(f"     - Разница в R²: {max(huber_metrics['r2']) - huber_metrics['r2'][idx_closest_to_2]:.4f}")
r2_loss = (max(huber_metrics['r2']) - huber_metrics['r2'][idx_closest_to_2]) / max(huber_metrics['r2']) * 100
print(f"     - Потери качества: {r2_loss:.2f}%")

print(f"\n 4. Интерпретация результатов:")
print("-" * 60)
print("   • Выбросы существенно ухудшают качество линейной регрессии")
print("   • HuberRegressor демонстрирует бóльшую стабильность коэффициентов")
print("   • Оптимальное значение epsilon близко к заданному (2.0)")
print("   • При удалении выбросов обе модели показывают схожее высокое качество")

# Визуализация сравнения коэффициентов
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
models = ['Linear\n(with outliers)', 'Linear\n(clean)', 'Huber\n(with outliers)', 'Huber\n(clean)']
coefs_features = [
    model_lr.coef_[0], model_lr_clean.coef_[0],
    model_huber.coef_[0], model_huber_clean.coef_[0]
]
coefs_updates = [
    model_lr.coef_[1], model_lr_clean.coef_[1],
    model_huber.coef_[1], model_huber_clean.coef_[1]
]
coefs_support = [
    model_lr.coef_[2], model_lr_clean.coef_[2],
    model_huber.coef_[2], model_huber_clean.coef_[2]
]

x = np.arange(len(models))
width = 0.25

plt.bar(x - width, coefs_features, width, label='features', alpha=0.7)
plt.bar(x, coefs_updates, width, label='updates', alpha=0.7)
plt.bar(x + width, coefs_support, width, label='support', alpha=0.7)

plt.xlabel('Модель и условие', fontsize=12)
plt.ylabel('Значения коэффициентов', fontsize=12)
plt.title('Сравнение коэффициентов моделей', fontsize=14, fontweight='bold')
plt.xticks(x, models, rotation=15)
plt.legend()
plt.grid(alpha=0.3, axis='y')

plt.subplot(1, 2, 2)
# Сравнение стабильности
stability_data = {
    'Linear\nRegression': lr_coef_diff,
    'Huber\nRegressor': huber_coef_diff
}
plt.bar(stability_data.keys(), stability_data.values(), color=['red', 'green'], alpha=0.7)
plt.ylabel('Среднее изменение коэффициентов', fontsize=12)
plt.title('Стабильность коэффициентов при удалении выбросов', fontsize=14, fontweight='bold')
plt.grid(alpha=0.3, axis='y')

# Добавляем значения
for i, (model, value) in enumerate(stability_data.items()):
    plt.text(i, value + 0.01, f'{value:.4f}', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
````

##  Ответы на контрольные вопросы

1. Почему MSE более чувствительна к выбросам, чем MAE? Приведите математическое обоснование через производные.

Ответ: В MSE производная линейно зависит от величины ошибки (чем больше ошибка, тем больше градиент), а MAE производная константна  независимо от величины ошибки.

2. Как параметр `epsilon` в Huber loss влияет на поведение модели? При каких значениях она ведёт себя как MSE, а при каких — как MAE?

Ответ: Параметр `epsilon` в Huber loss регулирует чувствительность к выбросам. При  больших  значениях она ведёт себя как MSE, а при маленьких — как MAE.

3. Почему масштабирование признаков важно для SGDRegressor? Что произойдёт, если признаки имеют разный масштаб?

Ответ: Масштабирование необходимо для того чтобы все признаки обучались с одинаковой скоростью.Если признаки имеют разный масштаб то, один и тот же шаг обучения может быть слишком большим для маленьких признаков и слишком маленьким для больших.

4. В каких практических задачах предпочтительнее использовать Huber loss вместо MSE? Приведите 2-3 примера.

Ответ: В практических задачах предпочтительнее использовать Huber loss вместо MSE если в данных есть выбросы, которые нужно учитывать, но не позволять им доминировать над моделью.

5. Как интерпретировать отрицательное значение R²? О чём это говорит о модели?

Ответ: Отрицательный R² означает, что модель хуже, чем простое предсказание среднего значения.Это говорит о том что модель плоха.

6. Почему при увеличении learning rate модель может "разойтись"? Проиллюстрируйте на примере формулы обновления весов.

Ответ:Из-за слишком большого шага мы не попадаем в точку минимума, а перелетаем её, и на каждой следующей итерации отклоняемся всё сильнее.
Пример:Формула обновления веса:
Новая гиря = Старая гиря — learning rate × Ошибка
Допустим, правильный вес (истина) — 10 кг, а сейчас ошиблись и поставили гирю 5 кг. Ошибка = 5 кг.

1.Нормальный learning rate:
Мы добавляем 1 кг (learning rate маленький).
5 + 1 × 5 = 10 кг. (равновесие)

2.Большой learning rate (переборщили):
Мы добавляем сразу по 3 кг за раз.
5 + 3 × 5 = 20 кг. (переложили) Теперь весы ушли в другую сторону. Ошибка стала -10 кг.

3.Модель "расходится" (весы ломаются):
исправить перевес (20 кг) (learning rate = 3).

Считаем: 20 — 3 × (-10) = 20 + 30 = 50 кг.
Итог: Весы не могут найти равновесие, потому что каждый раз мы добавляем или убираем слишком много массы

8. В чём разница между стохастическим, мини-батч и пакетным градиентным спуском? Когда какой использовать?

Ответ:Стохастический (SGD): Берет один случайный пример, сразу считает его ошибку и делает шаг. Очень быстрый, но шаги получаются "дергаными" (шумными).Мини-батч (Mini-batch GD): Золотая середина. Делит данные на небольшие пачки (например, по 32 или 64 примера).Пакетный (Batch GD): Смотрит сразу на все обучающие примеры, вычисляет общую ошибку и только потом делает один шаг.
Когда что использовать:
Пакетный: Только если данных очень мало (тысячи строк), или для отладки.
Стохастический: Когда данные поступают потоком в реальном времени (онлайн-обучение) и нет возможности загрузить всё сразу.
Мини-батч: Используется в 99% случаев.
