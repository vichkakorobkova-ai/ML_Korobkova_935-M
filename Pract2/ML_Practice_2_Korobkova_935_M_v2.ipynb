## 1. Введение в EDA (Exploratory Data Analysis)
Исследовательский анализ данных — это критический этап перед построением моделей. По оценкам индустрии, до 80% времени проекта уходит на подготовку и понимание данных. Цель EDA — не просто построить графики, а сформулировать гипотезы, обнаружить аномалии и понять структуру данных.

**Ключевые вопросы EDA:**
1.  **Качество данных:** Есть ли пропуски? Есть ли выбросы?
2.  **Распределения:** Нормальное ли распределение у признаков? Есть ли перекос (skewness)?
3.  **Зависимости:** Какие признаки коррелируют с целевой переменной? Есть ли мультиколлинеарность?
4.  **Типы данных:** Какие признаки категориальные, какие численные?

## 2. Теоретический базис методов

### 2.1. Пропущенные значения (Missing Values)
*   **MCAR (Missing Completely At Random):** Пропуск не зависит ни от чего.
*   **MAR (Missing At Random):** Пропуск зависит от других наблюдаемых переменных.
*   **MNAR (Missing Not At Random):** Пропуск зависит от самого пропущенного значения.
*   **Стратегии заполнения:**
    *   *Среднее (Mean):* Для нормальных распределений. Чувствительно к выбросам.
    *   *Медиана (Median):* Для распределений с выбросами.
    *   *Мода (Mode):* Для категориальных данных.

### 2.2. Выбросы (Outliers)
*   **IQR (Interquartile Range):** Метод межквартильного размаха. Выбросы за пределами $[Q1 - 1.5 \cdot IQR, Q3 + 1.5 \cdot IQR]$. Не требует предположения о нормальности.
*   **Z-Score:** Количество стандартных отклонений от среднего. Обычно порог $|Z| > 3$. Работает хорошо для нормальных распределений.

### 2.3. Преобразование признаков
*   **Логарифмирование:** Уменьшает правосторонний перекос (right skew).
*   **Нормализация (MinMax):** Приводит к диапазону $[0, 1]$.
*   **Стандартизация (Z-score):** Приводит к $\mu=0, \sigma=1$.

## 📋 Справочная таблица: Методы корреляции

| Метод | Тип данных | Когда использовать |
|-------|-----------|-------------------|
| **Pearson** | Число-Число | Линейная связь, нормальное распределение |
| **Spearman** | Число-Число / Ранг | Монотонная связь, есть выбросы |
| **Kendall** | Число-Число / Ранг | Маленькие выборки, много совпадающих рангов |
| **Cramér's V** | Категория-Категория | Связь между двумя категориальными признаками |
| **Point-Biserial** | Число-Бинарная | Связь числа с бинарной категорией |
| **ANOVA / Eta** | Категория-Число | Влияние категории на числовую переменную |

```python
# Импорт библиотек
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer

#Источник
#Dua, D., & Graff, C. (2019). UCI Machine Learning Repository [http://archive.ics.uci.edu/ml].
#Irvine, CA: University of California, School of Information and Computer Science.
# Загрузка датасета (red wine)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, sep=';')

# Вывод первых 3 строк и информации о данных
print("\n" + "="*100)
print("Вывод первых 3 строк и информации о данных")
print("="*100)
print("Первые 3 строк датасета:")
print(df.head(3))
print("\nИнформация о данных:")
print(df.info())
print("\nОписательная статистика исходных данных:")
print(df.describe())


# === === === === === === === === === === === ====== === === === === === === === === === === === === === === === === === === === === === === ===== === === === === === === === === === === ===
# ШАГ 1: Создание пропусков в признаке alcohol
# === === === === === === === === === === === === === === === === === === === === === === === ====== === === === === === === === === === === === === === === === === === === === === === === ===

np.random.seed(42)                                                                                   # Установим seed для воспроизводимости

df_missing = df.copy()                                                                                # Копия исходного датафрейма
missing_indices = np.random.choice(df_missing.index, size=int(0.15 * len(df_missing)), replace=False) # Индексы, которые станут пропусками

# Создаем пропуски в столбце 'alcohol'
df_missing.loc[missing_indices, 'alcohol'] = np.nan

print("\n" + "="*100)
print("Пункт 1: Создание пропусков в признаке alcohol")
print("="*100)
print(f"\nСоздано пропусков: {df_missing['alcohol'].isna().sum()} (15% от данных)")
print("\nПервые 3 строк с пропусками:")
print(df_missing.head(3))

# === === === === === === === === === === === ==== === === === ===
# ШАГ 2 : Заполните пропуски двумя разными способами
# === === === === === === === === === === === ==== === === === ===
print("\n" + "="*100)
print("Пункт 2: Заполните пропуски двумя разными способами ")
print("="*100)
# ==== === === === === === === ==== === === === ===
# ШАГ 2.1 : Заполнение средним значением (Mean Imputation)
# === === === === === === === === === === === ==== === === === ===
print("\n" + "="*70)
print("2.1 Заполнение средним значением (Mean Imputation)")
print("="*70)
# Заполнение средним значением (Mean Imputation)
df_mean = df_missing.copy()
mean_imputer = SimpleImputer(strategy='mean')
df_mean['alcohol'] = mean_imputer.fit_transform(df_mean[['alcohol']])
print("\nЗаполнение средним выполнено.")
print(f"Среднее значение после импутации: {df_mean['alcohol'].mean():.4f}")
print(f"Стандартное отклонение после импутации: {df_mean['alcohol'].std():.4f}")

# === === === === === === === === === === === ==== === === === ===
# ШАГ 2.2 : Заполнение медианой (Median Imputation)
# === === === === === === === === === === === ==== === === === ===
# Заполнение медианой (Median Imputation)
df_median = df_missing.copy()
median_imputer = SimpleImputer(strategy='median')
df_median['alcohol'] = median_imputer.fit_transform(df_median[['alcohol']])
print("\n" + "="*70)
print("2.2 Заполнение медианой (Median Imputation)")
print("="*70)
print("\nЗаполнение медианой выполнено.")
print(f"Среднее значение после импутации: {df_median['alcohol'].mean():.4f}")
print(f"Стандартное отклонение после импутации: {df_median['alcohol'].std():.4f}")


# === === === === === === === === === === === ==== === === === === === === === === === === === === === === === ==== === === === ===
# ШАГ 3 : Постройте гистограммы распределения признака до и после каждого метода.
# === === === === === === === === === === === ==== === === === ====== === === === === === === === === === === ==== === === === ===
print("\n" + "="*100)
print("Пункт 3: Постройте гистограммы распределения признака до и после каждого метода ")
print("="*100)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. ДО заполнения (данные с пропусками)
axes[0].hist(df_missing['alcohol'].dropna(), bins=30, edgecolor='black', alpha=0.7, color='salmon')
axes[0].axvline(df_missing['alcohol'].dropna().mean(), color='red', linestyle='--', linewidth=2,
                label=f"Среднее: {df_missing['alcohol'].dropna().mean():.2f}")
axes[0].set_title('ДО заполнения (с пропусками)')
axes[0].set_xlabel('alcohol')
axes[0].set_ylabel('Частота')
axes[0].legend()
axes[0].grid(axis='y', linestyle='--', alpha=0.3)

# 2. ПОСЛЕ заполнения СРЕДНИМ
axes[1].hist(df_mean['alcohol'], bins=30, edgecolor='black', alpha=0.7, color='lightgreen')
axes[1].axvline(df_mean['alcohol'].mean(), color='red', linestyle='--', linewidth=2,
                label=f"Среднее: {df_mean['alcohol'].mean():.2f}")
axes[1].set_title('ПОСЛЕ заполнения (среднее)')
axes[1].set_xlabel('alcohol')
axes[1].set_ylabel('Частота')
axes[1].legend()
axes[1].grid(axis='y', linestyle='--', alpha=0.3)

# 3. ПОСЛЕ заполнения МЕДИАНОЙ
axes[2].hist(df_median['alcohol'], bins=30, edgecolor='black', alpha=0.7, color='gold')
axes[2].axvline(df_median['alcohol'].mean(), color='red', linestyle='--', linewidth=2,
                label=f"Среднее: {df_median['alcohol'].mean():.2f}")
axes[2].set_title('ПОСЛЕ заполнения (медиана)')
axes[2].set_xlabel('alcohol')
axes[2].set_ylabel('Частота')
axes[2].legend()
axes[2].grid(axis='y', linestyle='--', alpha=0.3)

plt.suptitle('Распределение признака alcohol: до и после заполнения пропусков', fontsize=12)
plt.tight_layout()
plt.show()

# === === === === === === === === === === === ==== === === === ====== === ====== === === === === === === === === === === ==== === === === ===
# ШАГ 4 :Рассчитайте, как изменилось среднее и стандартное отклонение.
# === === === === === === === === === === === ==== === === === ====== === ====== === === === === === === === === === === ==== === === === ===
# --- СОЗДАЁМ СЛОВАРЬ СО СТАТИСТИКАМИ (ДОБАВЬТЕ ЭТОТ БЛОК!) ---
stats = {
    'Метод': ['Исходные', 'С пропусками', 'Заполнение средним', 'Заполнение медианой'],
    'Среднее': [
        df['alcohol'].mean(),
        df_missing['alcohol'].mean(),
        df_mean['alcohol'].mean(),
        df_median['alcohol'].mean()
    ],
    'Станд. отклонение': [
        df['alcohol'].std(),
        df_missing['alcohol'].std(),
        df_mean['alcohol'].std(),
        df_median['alcohol'].std()
    ]
}
stats_df = pd.DataFrame(stats) # Создаем DataFrame из словаря со статистиками

# Рассчитываем процент изменения среднего относительно исходных данных (строка 0)
stats_df['Изменение среднего (%)'] = ((stats_df['Среднее'] - stats_df.loc[0, 'Среднее']) / stats_df.loc[0, 'Среднее'] * 100).round(2)
# Рассчитываем процент изменения стандартного отклонения относительно исходных данных                                                                                                                                    # Формула: (новое - исходное) / исходное * 100%
stats_df['Изменение станд. откл. (%)'] = ((stats_df['Станд. отклонение'] - stats_df.loc[0, 'Станд. отклонение']) / stats_df.loc[0, 'Станд. отклонение'] * 100).round(2)

print("\n" + "="*100)
print("СРАВНЕНИЕ МЕТОДОВ")
print("="*100)
print(stats_df.to_string(index=False))

print("="*100)
print("ВЫВОДЫ")
# Находим метод с минимальными искажениями
abs_mean_change = stats_df.iloc[1:, 3].abs()  # абсолютное изменение среднего для методов
abs_std_change = stats_df.iloc[1:, 4].abs()   # абсолютное изменение станд. отклонения для методов

best_method_mean = stats_df.iloc[1:].loc[abs_mean_change.idxmin(), 'Метод']
best_method_std = stats_df.iloc[1:].loc[abs_std_change.idxmin(), 'Метод']

print(f"Метод, лучше всего сохранивший среднее значение: {best_method_mean}")
print(f"Метод, лучше всего сохранивший стандартное отклонение: {best_method_std}")
```

> **Подсказка:** Если выбросов > 5%, удаление может привести к потере информации. Рассмотрите возможность "обрезки" (capping) значений вместо удаления.

```python
# Импорт библиотек
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
import scipy.stats as stats

#Источник
#Dua, D., & Graff, C. (2019). UCI Machine Learning Repository [http://archive.ics.uci.edu/ml].
#Irvine, CA: University of California, School of Information and Computer Science.
# Загрузка датасета (red wine)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, sep=';')

# === === === === === === === === === === === ====== === === === === === === === === === === ===
# ШАГ 1: Выберите числовой признак
# === === === === === === === === === === === ====== === === === === === === === === === === ===
print("\n" + "="*100)
print("Пункт 1: Выбор числового признака")
print("="*100)

# Вычисление корреляции всех признаков с quality
correlations = df.corr()['quality'].sort_values(ascending=False)
print("Корреляция признаков с качеством вина (quality):")
print(correlations)

# Визуализация матрицы корреляций
plt.figure(figsize=(9, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Матрица корреляций признаков вина')
plt.tight_layout()
plt.show()

# === === === === === === === === === === === ====== === === ===
# ШАГ 2: НАЙДИТЕ ВЫБРОСЫ МЕТОДОМ IQR И МЕТОДОМ Z-SCORE.
# === === === === === === === === === === === ====== === === ===
print("\n" + "="*100)
print("Пункт 2: НАЙДИТЕ ВЫБРОСЫ МЕТОДОМ IQR И МЕТОДОМ Z-SCORE")
print("="*100)
# === === === === === === === === === === === ====== === === ===
# ШАГ 2.1: Найдите выбросы методом IQR
# === === === === === === === === === === === ====== === === ===
# Выбираем признак для анализа
feature = 'alcohol'
data = df[feature]

# Метод IQR
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

lower_bound_iqr = Q1 - 1.5 * IQR
upper_bound_iqr = Q3 + 1.5 * IQR

# Находим выбросы по методу IQR
outliers_iqr = data[(data < lower_bound_iqr) | (data > upper_bound_iqr)]

print("="*70)
print(" 2.1 МЕТОД IQR (Межквартильный размах)")
print("="*70)
print(f"Признак: {feature}")
print(f"Q1 (25-й процентиль): {Q1:.3f}")
print(f"Q3 (75-й процентиль): {Q3:.3f}")
print(f"IQR: {IQR:.3f}")
print(f"Нижняя граница (Q1 - 1.5*IQR): {lower_bound_iqr:.3f}")
print(f"Верхняя граница (Q3 + 1.5*IQR): {upper_bound_iqr:.3f}")
print(f"Количество выбросов: {len(outliers_iqr)}")
print(f"Процент выбросов: {len(outliers_iqr)/len(data)*100:.2f}%")
print(f"Значения выбросов: {sorted(outliers_iqr.values)}")

# === === === === === === === === === === === ====== === === ===
# ШАГ 2.2: Найдите выбросы методом Z-Score
# === === === === === === === === === === === ====== === === ===

# Метод Z-Score
z_scores = np.abs(stats.zscore(data))
threshold = 3  # стандартный порог для z-score
outliers_zscore = data[z_scores > threshold]

print("\n" + "="*70)
print(" 2.2 МЕТОД Z-SCORE")
print("="*70)
print(f"Признак: {feature}")
print(f"Порог Z-Score: {threshold}")
print(f"Количество выбросов: {len(outliers_zscore)}")
print(f"Процент выбросов: {len(outliers_zscore)/len(data)*100:.2f}%")
print(f"Значения выбросов: {sorted(outliers_zscore.values)}")


# === === === === === === === === === === === ====== === === ====== === === === === === === ====== === === ===
# ШАГ 3: Визуализуйте их на Boxplot и Scatterplot (подсветите выбросы красным цветом).
# === === === === === === === === === === === ====== === === ====== === === === === === === ====== === === ===

print("\n" + "="*100)
print("Пункт 3: Визуализация (выбросы красным цветом)")
print("="*100)

# Создаем копию данных для визуализации с подсветкой выбросов
df_plot = df.copy()
df_plot['is_outlier'] = False
df_plot.loc[outliers_iqr.index, 'is_outlier'] = True  # используем выбросы по IQR

# Создаем фигуру с двумя подграфиками
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# График 1: Boxplot
sns.boxplot(y=feature, data=df, ax=axes[0], color='lightblue')
axes[0].set_title(f'Boxplot признака {feature}')
axes[0].set_ylabel(feature)

# Добавляем точки выбросов на boxplot (альтернативный способ)
axes[0].scatter(x=[0]*len(outliers_iqr), y=outliers_iqr, color='red', s=50,
                alpha=0.6, label=f'Выбросы (n={len(outliers_iqr)})')
axes[0].legend()

# График 2: Scatterplot (используем другой признак для оси X, например, quality)
colors = ['red' if x else 'blue' for x in df_plot['is_outlier']]
axes[1].scatter(df['quality'], df[feature], c=colors, alpha=0.6, s=30)
axes[1].set_xlabel('quality')
axes[1].set_ylabel(feature)
axes[1].set_title(f'Scatterplot: quality vs {feature}')
axes[1].legend(['Нормальные данные', 'Выбросы'])

# Добавляем горизонтальные линии границ IQR
axes[1].axhline(y=lower_bound_iqr, color='green', linestyle='--', alpha=0.7, label='Нижняя граница IQR')
axes[1].axhline(y=upper_bound_iqr, color='green', linestyle='--', alpha=0.7, label='Верхняя граница IQR')
plt.tight_layout()
plt.show()

# === === === === === === === === === === === === === === ===
# ШАГ 4: Посчитайте процент выбросов от общего числа данных
# === === === === === === === === === === === === === === ===

# Анализ влияния выбросов на статистику
print("\n" + "="*70)
print("Пункт 4: АНАЛИЗ ВЛИЯНИЯ ВЫБРОСОВ")
print("="*70)

# Статистика до удаления выбросов
print("\nСтатистика ДО удаления выбросов:")
print(f"Среднее: {data.mean():.3f}")
print(f"Медиана: {data.median():.3f}")
print(f"Стандартное отклонение: {data.std():.3f}")

# Статистика после удаления выбросов
data_clean = data[~data.index.isin(outliers_iqr.index)]
print("\nСтатистика ПОСЛЕ удаления выбросов:")
print(f"Среднее: {data_clean.mean():.3f}")
print(f"Медиана: {data_clean.median():.3f}")
print(f"Стандартное отклонение: {data_clean.std():.3f}")

print(f"\nИзменение среднего: {(data_clean.mean() - data.mean()):.3f} ({(data_clean.mean()/data.mean() - 1)*100:.2f}%)")
print(f"Изменение стандартного отклонения: {(data_clean.std() - data.std()):.3f} ({(data_clean.std()/data.std() - 1)*100:.2f}%)")

```

**Цель:** Создать новый признак и проверить его полезность.
**Инструкция:**

> **Подсказка:** Хороший признак должен иметь физический смысл. Избегайте "мусорных" комбинаций.

```python
# Импорт библиотек
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
import scipy.stats as stats

#Источник
#Dua, D., & Graff, C. (2019). UCI Machine Learning Repository [http://archive.ics.uci.edu/ml].
#Irvine, CA: University of California, School of Information and Computer Science.
# Загрузка датасета (red wine)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, sep=';')

# === === === === === === === === === === === ====== === ===
# ШАГ 1: Создайте новый признак на основе существующих
# === === === === === === === === === === === ====== === ===

print("\n" + "="*200)
print("Пункт 1: Создание нового признака на основе существующего")
print("="*200)

# Исходные признаки и их корреляция с качеством
print("\nИсходные признаки и их корреляция с quality:")
corr_with_quality = df.corr()['quality'].sort_values(ascending=False)
print(corr_with_quality)

# Признак 1: Соотношение кислот (citric acid / volatile acidity)
# Гипотеза: Чем больше лимонной кислоты и меньше летучей кислоты, тем выше качество
df['acid_ratio'] = df['citric acid'] / (df['volatile acidity'] + 0.001)  # +0.001 чтобы избежать деления на 0

# Признак 2: Произведение алкоголя и сульфатов
# Гипотеза: Комбинация алкоголя и сульфатов (консервантов) может лучше предсказывать качество
df['alcohol_sulphates'] = df['alcohol'] * df['sulphates']

# Признак 3: Индекс насыщенности (density * residual sugar)
# Гипотеза: Плотность и остаточный сахар вместе определяют "тело" вина
df['body_index'] = df['density'] * df['residual sugar'] * 1000  # умножение на 1000 для масштабирования

# Признак 4: Общая кислотность (fixed acidity + volatile acidity + citric acid)
df['total_acidity'] = df['fixed acidity'] + df['volatile acidity'] + df['citric acid']


print("\n" + "-"*50)
print("Созданы новые признаки:")
print(f"1. acid_ratio = лимонная кислота / летучая кислотность")
print(f"2. alcohol_sulphates = алкоголь * сульфаты")
print(f"3. body_index = плотность * остаточный сахар * 1000")
print(f"4. total_acidity = фиксированная кислотность + летучая кислотность + лимонная кислота")


# === === === === === === === === === === === ====== === ====== === ===
# ШАГ 2: Постройте корреляционную матрицу с включением нового признака.
# === === === === === === === === === === === ====== === ====== === ===

print("\n" + "="*200)
print("Пункт 2: Корреляционная матрица с включением нового признака")
print("="*200)

# Вычисляем корреляцию всех признаков (включая новые) с quality
all_features = df.columns.tolist()
new_features = ['acid_ratio', 'alcohol_sulphates', 'body_index', 'total_acidity' ]

correlations = df.corr()['quality'].sort_values(ascending=False)
print("\nКорреляция всех признаков с quality (топ-15):")
print(correlations.head(15))

# Сравнение старых и новых признаков
print("\nСравнение исходных и новых признаков:")
print("-" * 100)
print(f"{'Признак':<25} {'Корреляция':<15} {'Тип'}")
print("-" * 100)

for feat in correlations.head(15).index:
    if feat in new_features:
        print(f"{feat:<25} {correlations[feat]:<15.4f} {' НОВЫЙ'}")
    elif feat == 'quality':
        print(f"{feat:<25} {correlations[feat]:<15.4f} {' ЦЕЛЕВОЙ'}")
    else:
        print(f"{feat:<25} {correlations[feat]:<15.4f} {' ИСХОДНЫЙ'}")

# Визуализация корреляционной матрицы
plt.figure(figsize=(14, 10))
# Берем топ-15 признаков для наглядности
top_features = correlations.head(15).index
sns.heatmap(df[top_features].corr(), annot=True, cmap='coolwarm', center=0,
            fmt='.2f', linewidths=0.5)
plt.title('Корреляционная матрица (топ-15 признаков, включая новые)', fontsize=14)
plt.tight_layout()
plt.show()

# === === === === === === === === === === === ====== === ====== === ===
# ШАГ 3: График зависимости нового признака от целевой переменной.
# === === === === === === === === === === === ====== === ====== === ===

print("\n" + "="*100)
print("Пункт 3: График зависимости нового признака от целевой переменной.")
print("="*100)

# Создаем графики для каждого нового признака (исправленная версия)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for i, feature in enumerate(new_features):
    # Исправленный boxplot - добавляем hue и legend=False
    sns.boxplot(x='quality', y=feature, data=df, ax=axes[i],
                palette='viridis', hue='quality', legend=False)
    axes[i].set_title(f'{feature}\nКорреляция с quality: {df[feature].corr(df["quality"]):.4f}')
    axes[i].set_xlabel('Качество вина')
    axes[i].set_ylabel(feature)

plt.tight_layout()
plt.show()

print("\n" + "="*100)
print("ВЫВОД")
print("="*100)

# Определяем лучший новый признак, если ещё не определён
if 'best_new_feature' not in locals():
    new_features = ['acid_ratio', 'alcohol_sulphates', 'body_index', 'total_acidity']
    best_new_feature = None
    best_new_corr = -1

    for feature in new_features:
        corr = df[feature].corr(df['quality'])
        if abs(corr) > abs(best_new_corr):
            best_new_corr = corr
            best_new_feature = feature
    print(f"Лучший новый признак: {best_new_feature} (корреляция: {best_new_corr:.4f})")

# Сравнение с лучшим исходным признаком
best_original = 'alcohol'
best_original_corr = df[best_original].corr(df['quality'])

print(f"\nЛучший исходный признак: {best_original}")
print(f"Корреляция с качеством (quality): {best_original_corr:.4f}")

print(f"\nЛучший новый признак: {best_new_feature}")
print(f"Корреляция с качеством (quality): {best_new_corr:.4f}")

```
